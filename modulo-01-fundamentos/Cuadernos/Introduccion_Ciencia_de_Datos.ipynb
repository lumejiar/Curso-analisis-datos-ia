
# Introducción a la Ciencia de Datos

Bienvenido al curso **Introducción a la Ciencia de Datos**.

La ciencia de datos es una disciplina que combina matemáticas, estadística, programación y conocimiento del dominio para transformar datos en información útil para la toma de decisiones.

Actualmente, empresas, gobiernos, investigadores y organizaciones generan enormes cantidades de datos. La capacidad de analizarlos permite descubrir patrones, realizar predicciones y automatizar procesos.



## ¿Por qué estudiar Ciencia de Datos?

La ciencia de datos permite:

- Analizar grandes volúmenes de información.
- Identificar tendencias y comportamientos.
- Construir modelos predictivos.
- Automatizar tareas de análisis.
- Apoyar la toma de decisiones basada en evidencia.

### Ejemplos de aplicación

- Predicción de ventas.
- Diagnóstico médico asistido por computador.
- Sistemas de recomendación.
- Detección de fraude.
- Análisis de redes sociales.
- Optimización de procesos industriales.


## Trabajando con datos

Los datos pueden almacenarse en listas, tablas o bases de datos. Empezaremos con una lista sencilla.


```

temperaturas = [25, 27, 24, 30, 28, 29, 26]

print("Temperaturas:", temperaturas)
print("Máxima:", max(temperaturas))
print("Mínima:", min(temperaturas))
print("Promedio:", sum(temperaturas)/len(temperaturas))
```


## Introducción a Pandas

Pandas es una de las bibliotecas más importantes para el análisis de datos.


```

import pandas as pd

datos = pd.DataFrame({
    "Nombre": ["Ana", "Juan", "María", "Carlos"],
    "Edad": [20, 25, 22, 30],
    "Ciudad": ["Cali", "Bogotá", "Medellín", "Cali"]
})

datos
```


## Estadística básica

La estadística es una herramienta fundamental para resumir y comprender datos.


```

datos_numericos = [12, 15, 18, 20, 25, 30, 35]

media = sum(datos_numericos)/len(datos_numericos)

print("Media:", media)
print("Máximo:", max(datos_numericos))
print("Mínimo:", min(datos_numericos))
```


## Visualización de datos

Una imagen puede revelar patrones difíciles de detectar en una tabla.


```

import matplotlib.pyplot as plt

ventas = [15, 18, 22, 30, 28, 35]

plt.plot(ventas)
plt.title("Ventas mensuales")
plt.xlabel("Mes")
plt.ylabel("Ventas")
plt.show()
```
